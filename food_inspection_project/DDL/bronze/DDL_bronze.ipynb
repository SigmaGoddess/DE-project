```sql
-- Creating MANAGED table from Landing with filters
DROP TABLE IF EXISTS food_inspection.bronze.properties_bronze;

CREATE TABLE food_inspection.bronze.properties_bronze
USING DELTA
AS
SELECT *
FROM food_inspection.landing.food_api
WHERE county IS NOT NULL
  AND inspection_type IS NOT NULL
  AND CAST(total_critical_violations AS INT) >= 0; 
```
