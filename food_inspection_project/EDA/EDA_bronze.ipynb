```sql
WITH total AS (
    SELECT COUNT(*) AS n
    FROM food_inspection.bronze.properties_bronze
), 
nullls AS (
SELECT 
    COUNT(*) - COUNT(:id)                          AS nulos_id,
    COUNT(*) - COUNT(address)                     AS nulos_address,
    COUNT(*) - COUNT(city)                         AS nulos_city,
    COUNT(*) - COUNT(county)                       AS nulos_county,
    COUNT(*) - COUNT(date)                         AS nulos_date,
    COUNT(*) - COUNT(description)                 AS nulos_description,
    COUNT(*) - COUNT(facility)                     AS nulos_facility,
    COUNT(*) - COUNT(facility_address)             AS nulos_facility_address,
    COUNT(*) - COUNT(municipality)                 AS nulos_municipality,
    COUNT(*) - COUNT(nys_health_operation_id)      AS nulos_nys_health_operation_id,
    COUNT(*) - COUNT(operation_name)               AS nulos_operation_name,
    COUNT(*) - COUNT(perm_operator_first_name)     AS nulos_perm_operator_first_name,
    COUNT(*) - COUNT(perm_operator_last_name)      AS nulos_perm_operator_last_name,
    COUNT(*) - COUNT(permit_expiration_date)       AS nulos_permit_expiration_date,
    COUNT(*) - COUNT(permitted_corp_name)          AS nulos_permitted_corp_name,
    COUNT(*) - COUNT(total_critical_violations)    AS nulos_total_critical_violations,
    COUNT(*) - COUNT(total_noncritical_violations) AS nulos_total_noncritical_violations,
    COUNT(*) - COUNT(violations)                   AS nulos_violations,
    COUNT(*) - COUNT(zip_code)                     AS nulos_zip_code,
    COUNT(*) - COUNT(permitted_d_b_a)              AS nulos_permitted_d_b_a
FROM food_inspection.bronze.properties_bronze
)
SELECT 
    t.n as total_registros,
    n.*
FROM total t, nullls n;
```

### Percentages of invalid votes

```sql
WITH totales AS (
    SELECT COUNT(*) as total FROM food_inspection.bronze.properties_bronze
)
SELECT 

    ROUND((t.total - COUNT(city)) * 100.0 / t.total, 2) as pct_nulos_city,
    ROUND((t.total - COUNT(facility_address)) * 100.0 / t.total, 2) as pct_nulos_facility_address,
    ROUND((t.total - COUNT(perm_operator_first_name)) * 100.0 / t.total, 2) as pct_nulos_perm_operator_first_name,
    ROUND((t.total - COUNT(perm_operator_last_name)) * 100.0 / t.total, 2) as pct_nulos_perm_operator_last_name,
    ROUND((t.total - COUNT(permit_expiration_date)) * 100.0 / t.total, 2) as pct_nulos_permit_expiration_date,
    ROUND((t.total - COUNT(permitted_corp_name)) * 100.0 / t.total, 2) as pct_nulos_permitted_corp_name,
    ROUND((t.total - COUNT(zip_code)) * 100.0 / t.total, 2) as pct_nulos_zip_code,
    ROUND((t.total - COUNT(permitted_d_b_a)) * 100.0 / t.total, 2) as pct_nulos_permitted_d_b_a
FROM food_inspection.bronze.properties_bronze
CROSS JOIN totales t
GROUP BY t.total;
```

### Critical percentage of null votes (+50%)

```sql
WITH totales AS (
    SELECT COUNT(*) as total FROM food_inspection.bronze.properties_bronze
), 
pct_nulls AS (
SELECT 

    ROUND((t.total - COUNT(city)) * 100.0 / t.total, 2) as pct_nulos_city,
    ROUND((t.total - COUNT(facility_address)) * 100.0 / t.total, 2) as pct_nulos_facility_address,
    ROUND((t.total - COUNT(perm_operator_first_name)) * 100.0 / t.total, 2) as pct_nulos_perm_operator_first_name,
    ROUND((t.total - COUNT(perm_operator_last_name)) * 100.0 / t.total, 2) as pct_nulos_perm_operator_last_name,
    ROUND((t.total - COUNT(permit_expiration_date)) * 100.0 / t.total, 2) as pct_nulos_permit_expiration_date,
    ROUND((t.total - COUNT(permitted_corp_name)) * 100.0 / t.total, 2) as pct_nulos_permitted_corp_name,
    ROUND((t.total - COUNT(zip_code)) * 100.0 / t.total, 2) as pct_nulos_zip_code,
    ROUND((t.total - COUNT(permitted_d_b_a)) * 100.0 / t.total, 2) as pct_nulos_permitted_d_b_a
FROM food_inspection.bronze.properties_bronze
CROSS JOIN totales t
GROUP BY t.total
)

SELECT column, percentage_nulls
FROM pct_nulls
UNPIVOT (
    percentage_nulls FOR column IN (
        pct_nulos_city, pct_nulos_facility_address, pct_nulos_perm_operator_first_name, 
        pct_nulos_perm_operator_last_name, pct_nulos_permit_expiration_date,
        pct_nulos_permitted_corp_name, pct_nulos_zip_code, pct_nulos_permitted_d_b_a
    )
)
WHERE percentage_nulls > 50
ORDER BY percentage_nulls DESC;
```

### Descriptive Statistics for Numerical Variables (Outliers)

```sql
SELECT * FROM food_inspection.bronze.properties_bronze LIMIT 10;
```

```sql
CREATE OR REPLACE TEMPORARY VIEW properties_clean_1 AS
SELECT
    :id,
    address,
    city,
    county,
    date,
    description,
    facility,
    facility_address,
    municipality,
    nys_health_operation_id,
    operation_name,
    perm_operator_first_name,
    perm_operator_last_name,
    permit_expiration_date,
    permitted_corp_name,
    permitted_d_b_a,
    zip_code,
    violations,
    local_health_department,
    CASE
        WHEN total_crit_not_corrected RLIKE '^[^a-zA-Z]+$' THEN total_crit_not_corrected::double
        ELSE NULL
    END AS total_crit_not_corrected,
    CASE
        WHEN total_critical_violations RLIKE '^[^a-zA-Z]+$' THEN total_critical_violations::double
        ELSE NULL
    END AS total_critical_violations,
    CASE
        WHEN total_noncritical_violations RLIKE '^[^a-zA-Z]+$' THEN total_noncritical_violations::double
        ELSE NULL
    END AS total_noncritical_violations
FROM food_inspection.bronze.properties_bronze;
```

### Basic Statitistcs

```sql
-- statistics on the number of critical violations
SELECT 
    'total_critical_violations' as columna,
    COUNT(*) as cantidad,
    ROUND(MIN(total_critical_violations), 2) as crit_vio_min,
    ROUND(MAX(total_critical_violations), 2) as crit_vio_max,
    ROUND(AVG(total_critical_violations), 2) as crit_vio_avg,
    ROUND(PERCENTILE(total_critical_violations, 0.5), 2) as crit_vio_median,
    ROUND(PERCENTILE(total_critical_violations, 0.25), 2) as crit_vio_p25,
    ROUND(PERCENTILE(total_critical_violations, 0.75), 2) as crit_vio_p75
FROM properties_clean_1
WHERE total_critical_violations IS NOT NULL AND total_critical_violations > 0
ORDER BY cantidad DESC;
```

```sql
SELECT 
    'total_noncritical_violations' as columna,
    COUNT(*) as cantidad,
    ROUND(MIN(total_noncritical_violations), 2) as noncrit_vio_min,
    ROUND(MAX(total_noncritical_violations), 2) as noncrit_vio_max,
    ROUND(AVG(total_noncritical_violations), 2) as noncrit_vio_avg,
    ROUND(PERCENTILE(total_noncritical_violations, 0.5), 2) as noncrit_vio_median,
    ROUND(PERCENTILE(total_noncritical_violations, 0.25), 2) as noncrit_vio_p25,
    ROUND(PERCENTILE(total_noncritical_violations, 0.75), 2) as noncrit_vio_p75
FROM properties_clean_1
WHERE total_noncritical_violations IS NOT NULL AND total_noncritical_violations > 0

UNION ALL

SELECT 
    'total_crit_not_corrected' as columna,
    COUNT(*) as cantidad,
    ROUND(MIN(total_crit_not_corrected), 2) as crit_notcorr_vio_min,
    ROUND(MAX(total_crit_not_corrected), 2) as crit_notcorr_vio_max,
    ROUND(AVG(total_crit_not_corrected), 2) as crit_notcorr_vio_avg,
    ROUND(PERCENTILE(total_crit_not_corrected, 0.5), 2) as crit_notcorr_vio_median,
    ROUND(PERCENTILE(total_crit_not_corrected, 0.25), 2) as crit_notcorr_vio_p25,
    ROUND(PERCENTILE(total_crit_not_corrected, 0.75), 2) as crit_notcorr_vio_p75
FROM properties_clean_1
WHERE total_crit_not_corrected IS NOT NULL AND total_crit_not_corrected > 0
ORDER BY cantidad DESC; 
```

### Duplicates and invalid entries

```sql
WITH total AS (
    SELECT 
        COUNT(*) as n
    FROM properties_clean_1
),
problems AS (
    SELECT 
        -- total_noncritical_violations invalid
        COUNT(CASE WHEN total_noncritical_violations IS NULL OR total_noncritical_violations <= 0 THEN 1 END) AS total_noncritical_violations_invalido,
        -- total_critical_violations invalid
        COUNT(CASE WHEN total_critical_violations IS NULL OR total_critical_violations <= 0 THEN 1 END) AS total_critical_violations_invalido,
        -- total_crit_not_corrected invalid
        COUNT(CASE WHEN total_crit_not_corrected IS NULL OR total_crit_not_corrected <=0 THEN 1 END) AS total_crit_not_corrected_invalido
    FROM properties_clean_1
)
SELECT 
    t.n AS total_registros,
    -- Porcentaje de cada problema sobre el total
    ROUND(total_noncritical_violations_invalido * 100.0 / n, 2) AS porcentaje_total_noncritical_violations_invalido,
    ROUND(total_critical_violations_invalido * 100.0 / n, 2) AS porcentaje_total_critical_violations_invalido,
    ROUND(total_crit_not_corrected_invalido * 100.0 / n, 2) AS 
    porcentaje_total_crit_not_corrected_invalido
FROM total t, problems p
ORDER BY total_registros DESC
```

###Duplicates

```sql
WITH duplicados AS (
    SELECT 
        address, 
        facility,
        COUNT(*) AS veces
FROM properties_clean_1
GROUP BY address, facility
HAVING COUNT(*) > 1
)

SELECT 
    COUNT(*) as grupos_duplicados,
    SUM(veces) as total_registros_duplicados,
    SUM(veces - 1) as registros_extra_por_duplicacion
FROM duplicados;
```

```sql
WITH ejemplos_duplicados(
    SELECT 
       COUNT(*) as veces,
       address, 
       facility 
    FROM properties_clean_1
    GROUP BY address, facility
    HAVING COUNT(*) > 1 
)

SELECT * 
FROM ejemplos_duplicados
ORDER BY veces DESC LIMIT 10;
```

```sql
WITH stats AS (
    SELECT 
        PERCENTILE(total_critical_violations, 0.01) as p01_1,
        PERCENTILE(total_critical_violations, 0.99) as p99_1,
        PERCENTILE(total_noncritical_violations, 0.01) as p01_2,
        PERCENTILE(total_noncritical_violations, 0.99) as p99_2
    FROM properties_clean_1
    WHERE total_critical_violations > 0 AND total_noncritical_violations > 0
)

SELECT 
    P.total_critical_violations as total_critical_violations,
    p.total_noncritical_violations as total_noncritical_violations,
    CASE
        WHEN p.total_critical_violations > s.p99_1 THEN 'Muy alto (> P99)'
        WHEN p.total_critical_violations < s.p01_1 THEN 'Muy bajo (< P01)'
        ELSE 'Normal'
    END as tipo_outlier,
    COUNT(*) as cantidad
FROM properties_clean_1 p
CROSS JOIN stats s
WHERE p.total_critical_violations > 0 AND p.total_noncritical_violations > 0
GROUP BY P.total_critical_violations, p.total_noncritical_violations
ORDER BY tipo_outlier, cantidad DESC; 
```
